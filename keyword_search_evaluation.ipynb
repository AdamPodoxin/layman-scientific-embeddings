```python
from utils import load_finetuned_qwen, round_down_to_2_decimals
from keyword_search_evaluation import get_scores
```

```python
def print_model_scores(model):
    scores = get_scores(model)
    print(f"Mean reciprocal rank: {round_down_to_2_decimals(scores["mean_reciprocal_rank"]):.2f}")
    print(f"Recall@15: {round_down_to_2_decimals(scores["recall_at_15"]):.2f}")
    print(f"Precision@15: {round_down_to_2_decimals(scores["precision_at_15"]):.2f}")
    print(f"F1@15: {round_down_to_2_decimals(scores["f1_at_15"]):.2f}")
```

Base scibert

```python
print_model_scores("allenai/scibert_scivocab_uncased")
```

Vanilla scibert

```python
print_model_scores("models/vanilla-scibert")
```

Jargon-layman scibert

```python
print_model_scores("models/jargon-layman-scibert-1.00")
```

Vanilla qwen

```python
print_model_scores(
    load_finetuned_qwen("models/vanilla-qwen")
)
```

Jargon-layman qwen

```python
print_model_scores(
    load_finetuned_qwen("models/jargon-layman-qwen")
)
```
