```python
def print_model_scores(model):
    scores = get_scores(model)
    print(f"Full match score: {scores["perfect_match_score"] * 100:.2f}%")
    print(f"Related keywords score: {scores["in_top_results_score"] * 100:.2f}%")
```

Base scibert

```python
print_model_scores("allenai/scibert_scivocab_uncased")
```

Vanilla scibert

```python
print_model_scores("models/vanilla-scibert")
```

Jargon-layman scibert

```python
print_model_scores("models/jargon-layman-scibert-1.00")
```

Vanilla qwen

```python
print_model_scores(
    load_finetuned_qwen("models/vanilla-qwen")
)
```

Jargon-layman qwen

```python
print_model_scores(
    load_finetuned_qwen("models/jargon-layman-qwen")
)
```
